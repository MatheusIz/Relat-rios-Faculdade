#Relatorio: Sistemas de Gestão de Notas de Alunos


```python
Nota_1 = int(input("Digiote a Nota: "))
Nota_2 = int(input("Digiote a Nota: "))
Nota_3 = int(input("Digiote a Nota: "))
Nota_4 = int(input("Digiote a Nota: "))
```

```python
Media = (Nota_1 + Nota_2 + Nota_3 + Nota_4) / 4
```

```python
if Media >= 7:
  Situação = "Aprovado"
elif Media <= 5:
  Situação = "Reprovado"
```

```python
print(Situação)
```
