```python
import pandas as pd
import sqlite3

# Conectando ao banco de dados
conexao = sqlite3.connect('dados_vendas.db')

# Lendo os dados da tabela vendas1 para um DataFrame do pandas
df_vendas = pd.read_sql_query("SELECT * FROM vendas1", conexao)

# Exibindo as primeiras 5 linhas do DataFrame
display(df_vendas.head())
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Configuração estética do Seaborn para gráficos.
sns.set_theme(style="whitegrid")

# Convert 'data_venda' to datetime and extract 'Mes'
df_vendas['data_venda'] = pd.to_datetime(df_vendas['data_venda'])
df_vendas['Mes'] = df_vendas['data_venda'].dt.month

# Create 'Faturamento' column from 'valor_venda'
df_vendas['Faturamento'] = df_vendas['valor_venda']

# Agrupando os dados por mês.
vendas_mensais = df_vendas.groupby('Mes')['Faturamento'].sum().reset_index()

plt.figure(figsize=(10, 5))
sns.lineplot(data=vendas_mensais, x='Mes', y='Faturamento', marker='o', color='b', linewidth=2.5)

plt.title('Tendência de Faturamento Mensal', fontsize=14, fontweight='bold')
plt.xlabel('Mês', fontsize=12)
plt.ylabel('Faturamento (R$)', fontsize=12)
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

```
