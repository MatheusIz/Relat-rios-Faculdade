#Relatorio BIBLIOTECAS E MÓDULOS EM PYTHON

```python
import matplotlib.pyplot as plt

class Livro:
    def __init__(self, titulo, autor, ano, genero):
        self.titulo = titulo
        self.autor = autor
        self.ano = ano
        self.genero = genero
        self.Cadastrado = False

    def exibir_detalhes(self):
        status = "Sim" if self.Cadastrado else "Não"
        print(f"📖 Título: {self.titulo}")
        print(f"✍️ Autor: {self.autor}")
        print(f"📅 Ano: {self.ano}")
        print(f"📚 Gênero: {self.genero}")
        print(f"✅ Cadastrado: {status}")

    def marcar_Cadastrado(self):
        self.Cadastrado = True
        print(f"✅ Livro '{self.titulo}' cadastrado com sucesso!")

    # Método para gerar o gráfico do progresso do livro usando Matplotlib
    def gerar_grafico_progresso(self):
        if self.Cadastrado:
            cadastrado_val = 1
            nao_cadastrado_val = 0
        else:
            cadastrado_val = 0
            nao_cadastrado_val = 1

        # Dados para o gráfico de pizza
        categorias = ['Cadastrado', 'Não Cadastrado']
        valores = [cadastrado_val, nao_cadastrado_val]
        cores = ['#4CAF50', '#FFC107']

        # Criando o gráfico
        plt.figure(figsize=(5, 5))
        plt.pie(valores, labels=categorias, autopct='%1.1f%%', startangle=90, colors=cores)
        plt.title(f"Status de Cadastro do Livro: {self.titulo}")
        plt.show()
```

```python
Livro1 = ("Retorno do Rei", "J. R. R. Tolkien", "1955", "Fantasia")
Livro2 = ("O Senhor dos Anéis", "J. R. R. Tolkien", "1954", "Fantasia")
Livro3 = ("O Hobbit", "J. R. R. Tolkien", "1937", "Fantasia")
Livro4 = ("Pequeno principe", "Antoine de Saint-Exupéry", "1943", "Infantil")
Livro5 = ("Harry Potter e a Pedra Filosofal", "J. K. Rowling", "1997", "Fantasia")
```

```python
# Criando uma lista vazia para armazenar o cadastro dos livros
cadastro_de_livros = []

# Cadastrando (instanciando) os livros
livro1 = Livro("Retorno do Rei", "J. R. R. Tolkien", "1955", "Fantasia")
livro2 = Livro("O Senhor dos Anéis", "J. R. R. Tolkien", "1954", "Fantasia")
livro3 = Livro("O Hobbit", "J. R. R. Tolkien", "1937", "Fantasia")
livro4 = Livro("Pequeno principe", "Antoine de Saint-Exupéry", "1943", "Infantil")
livro5 = Livro("Harry Potter e a Pedra Filosofal", "J. K. Rowling", "1997", "Fantasia")

cadastro_de_livros.append(livro1)
cadastro_de_livros.append(livro2)
cadastro_de_livros.append(livro3)
cadastro_de_livros.append(livro4)
cadastro_de_livros.append(livro5)

# Alterando o status de um dos livros
livro1.marcar_Cadastrado()
livro2.marcar_Cadastrado()
livro3.marcar_Cadastrado()
livro4.marcar_Cadastrado()
livro5.marcar_Cadastrado()
print() # Linha em branco para organizar a saída

# Listando todos os livros cadastrados
print("--- LIVROS CADASTRADOS ---")
for livro in cadastro_de_livros:
    livro.exibir_detalhes()
```

```python
import matplotlib.pyplot as plt

# 1. Criando uma lista de livros para testar o gráfico
# (Note que adicionamos os atributos: titulo, autor, ano, genero)
cadastro_de_livros = [
  Livro("Retorno do Rei", "J. R. R. Tolkien", "1955", "Fantasia"),
  Livro("O Senhor dos Anéis", "J. R. R. Tolkien", "1954", "Fantasia"),
  Livro("O Hobbit", "J. R. R. Tolkien", "1937", "Fantasia"),
  Livro("Pequeno principe", "Antoine de Saint-Exupéry", "1943", "Infantil"),
  Livro("Harry Potter e a Pedra Filosofal", "J. K. Rowling", "1997", "Fantasia")
]

# 2. Função para contar os gêneros e gerar o gráfico
def gerar_grafico_por_genero(lista_livros):
    # Dicionário para armazenar a contagem de cada gênero
    contagem_generos = {}

    # Percorre a lista acumulando a quantidade de livros por gênero
    for livro in lista_livros:
        genero = livro.genero
        if genero in contagem_generos:
            contagem_generos[genero] += 1
        else:
            contagem_generos[genero] = 1

    # Separando os dados para o Matplotlib
    generos = list(contagem_generos.keys())
    quantidades = list(contagem_generos.values())

    # --- CONFIGURAÇÃO DO GRÁFICO (MATPLOTLIB) ---
    plt.figure(figsize=(8, 5))

    # Criando o gráfico de barras verticais com uma cor amigável
    plt.bar(generos, quantidades, color='#34495e', width=0.5)

    # Customizando o visual
    plt.title("Quantidade de Livros por Gênero", fontsize=14, fontweight='bold')
    plt.xlabel("Gêneros Literários", fontsize=12)
    plt.ylabel("Número de Livros", fontsize=12)

    # Força o eixo Y a mostrar apenas números inteiros (1, 2, 3...)
    plt.yticks(range(0, max(quantidades) + 2))
    plt.grid(axis='y', linestyle='--', alpha=0.5)

    # Exibe o gráfico na tela do Colab
    plt.show()

# 3. Chamando a função para ver o resultado diante do código
gerar_grafico_por_genero(cadastro_de_livros)
```
